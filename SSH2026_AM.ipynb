````python
import os

file_name = 'my_message.txt'
message = 'これはテストメッセージです。'

# ファイルにメッセージを書き込む
with open(file_name, 'w', encoding='utf-8') as f:
    f.write(message)

# ファイルの絶対パスを取得して表示する
file_path = os.path.abspath(file_name)
print(f"メッセージは '{file_path}' に書き込まれました。")
````

````python
import numpy as np
import matplotlib.pyplot as plt

# xの値を生成
x = np.linspace(-10, 10, 100)

# y = 2x + 1
y = 2 * x + 1

# グラフを描画
plt.plot(x, y, label='y = 2x + 1')

# 軸とグリッド
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.grid(True)

# タイトルと凡例
plt.title('Graph of y = 2x + 1')
plt.legend()

# 表示
plt.show()

````

````python
import numpy as np
import matplotlib.pyplot as plt

# xの値を生成
x = np.linspace(-4, 2, 100)

# y = 2x + 1
y = x ** 2 + 2 * x + 2

# グラフを描画
plt.plot(x, y, label='y  = x ** 2 + 2 * x + 2')

# 軸とグリッド
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.grid(True)

# タイトルと凡例
plt.title('Graph of y =  x ** 2 + 2 * x + 2')
plt.legend()

# 表示
plt.show()
````
