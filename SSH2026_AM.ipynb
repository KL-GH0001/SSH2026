#### 関係演算子

#### 条件分岐
if else:文は条件式の値がTrueのとき，それに続くブロックを実行する．Ｆａｌｓｅ のとき，elseに続くブロックを実行する．ブロックとは，段付け(indent)された文の集まりのことをいう．

```python
x = 3
if x % 2 == 0:
  print(x, "is even")
else:
  print(x, "is odd")
```

#### 繰返し
「for 変数in 列」は繰返しを行う．range(5)は0,1,2,3,4 という列を表す．列の各要素を変数へ順に代入し，それに続くブロックを実行する．

```python
for i in range(5):
  print(i)
```

##### 総和の計算
「for 変数in 列」は繰返しを行う．range(1,11)は1，2, …,10という列を表す．列の各要素を変数へ順に代入し，それに続くブロックを実行する．この例では，totalへ初期値0を代入する．その後､ 1,2,3,…,10を順に加える．その結果 totalの値は55となる．



```python
total = 0

for i in range(1,11):
    total += i

print(total)
```

#### numpy 
Numpy(ナムパイ) は、Pythonで数値計算を高速に行うためのライブラリである. グラフを描く際のデータ作成に利用される.
linspace(L, R, n)は, 区間[L,R]をn等分する点列を生成する．

```python
import numpy as np

x = np.linspace(0, 1, 11)
y = 2 * x + 1
print(x)
print(y)
```

グラフ(1次関数)

```python
import numpy as np
import matplotlib.pyplot as plt

# xの値を生成
x = np.linspace(-10, 10, 100)

# y = 2x + 1
y = 2 * x + 1

# グラフを描画
plt.plot(x, y, label='y = 2x + 1')

# 軸とグリッド
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.grid(True)

# タイトルと凡例
plt.title('Graph of y = 2x + 1')
plt.legend()

# 表示
plt.show()

```

グラフ(2次関数)

```python
import numpy as np
import matplotlib.pyplot as plt

# xの値を生成
x = np.linspace(-4, 2, 100)

# y = 2x + 1
y = x ** 2 + 2 * x + 2

# グラフを描画
plt.plot(x, y, label='y  = x ** 2 + 2 * x + 2')

# 軸とグリッド
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.grid(True)

# タイトルと凡例
plt.title('Graph of y =  x ** 2 + 2 * x + 2')
plt.legend()

# 表示
plt.show()
```
